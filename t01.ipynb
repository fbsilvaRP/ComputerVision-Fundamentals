```python
from urllib.request import urlopen
import cv2
import numpy
import matplotlib.pyplot as plt

# --- 1.Obtendo uma imagem de uma página web

address = "https://denmartins.github.io/visao-computacional-book/_images/2f7d9afde91794072aa66b169257c785ade17f97dcc2ab0141a4f6faba54f64f.png"
resp = urlopen(address)                                                 #Abre a conexão de rede e faz a requisição HTTP GET
print(resp.status)                                                      #Verifica o status da resposta (200 significa "sucesso")

dados_brutos = resp.read()                                                #Baixa os bytes puros do arquivo de imagem (formato imutável)
imagem_Bytearray = bytearray(dados_brutos)                                #Converte os bytes em uma sequência mutável (lista de bytes)

imagem = numpy.asarray(imagem_Bytearray, dtype = "uint8")               #Transforma a lista de bytes em uma matriz unidimensional do NumPy
imagem = cv2.imdecode(imagem, cv2.IMREAD_COLOR)                         #Reconstrói a imagem(reorganiza os bytes em largura, altura e canais de cor)
imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)                        #Converte o padrão de cor do OpenCV (BGR) para o padrão do Matplotlib (RGB)

# ---
# --- 2.Exibindo a imagem (em um gráfico)
plt.imshow(imagem)
plt.show()





```
